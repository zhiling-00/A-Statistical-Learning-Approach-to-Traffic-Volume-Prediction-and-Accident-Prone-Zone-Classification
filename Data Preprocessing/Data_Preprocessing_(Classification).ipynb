```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder

# Load Dataset
from google.colab import files
uploaded = files.upload()
file = pd.read_csv('synthetic_road_accidents_100k.csv')
data_org = pd.DataFrame(file)
data_org.head()
```

## **Data Cleaning**

```python
# Remove Duplicate Data
data = data_org.drop_duplicates()
print((len(data_org)-len(data))*100/len(data_org) , '% of the rows in the dataset are duplicate and had been removed.')
```

```python
# Identify Missing Values
print("Missing Values in Dataset:")
print(data.isnull().sum())
```

### **Feature Engineering (For Testing Enhancement)**

# **Data Transformation using Label Encoding**

```python
# Identify Features Type
categorical_features = ['road_type', 'lighting', 'weather', 'time_of_day']
boolean_features = ['road_signs_present', 'public_road', 'holiday', 'school_season']
target_col = 'accident_risk'
id_col = 'id'
```

```python
# Get all feature columns (exclude id and target)
feature_cols = [col for col in data.columns if col not in [id_col, target_col]]

print(f"Feature Summary:")
print(f"Total features: {len(feature_cols)}")
print(f"Categorical: {len(categorical_features)}")
print(f"Boolean: {len(boolean_features)}")
print(f"Numerical: {len(feature_cols) - len(categorical_features) - len(boolean_features)}")
```

```python
# Convert Boolean Features to 0/1
for col in boolean_features:
    data.loc[:, col] = data[col].astype(int)
```

```python
# Label Encode Categorical Features
label_encoders = {}

for col in categorical_features:
    le = LabelEncoder()
    data.loc[:, col] = le.fit_transform(data[col])
    label_encoders[col] = le

    # Display Categorical Column and Their Unique Label
    print(f"Mapping for column '{col}':")
    for class_index, class_label in enumerate(le.classes_):
        print(f"{class_label}: {class_index}")
    print("\n")
```

```python
# Categorize target "accident risk" to Categorical Label
bins = [0, 0.25, 0.6, 1.0]
labels = ['Low', 'Medium', 'High']

if data['accident_risk'].dtype == 'category':
    data['accident_risk'] = data_org.loc[data.index, 'accident_risk']

# Convert column to categorical
data['accident_risk'] = pd.cut(
    data['accident_risk'],
    bins=bins,
    labels=labels,
    include_lowest=True
)
print(data['accident_risk'].value_counts())
# Label Encode
le_target = LabelEncoder()
y = le_target.fit_transform(data['accident_risk'])
print("Target classes:", list(le_target.classes_))
```
